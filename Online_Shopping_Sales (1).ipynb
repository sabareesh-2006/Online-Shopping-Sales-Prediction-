1.Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score)
```

2.Loading the Dataset

```python
df=pd.read_csv("online_shopping_sales_100_rows_20_columns.csv")
print(df.head())
```

3.Data Analysis



```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nSummary Statistics:")
print(df.describe())

print("\nColumn Names:")
print(df.columns)
```

4.Identifying Missing Values

```python
print("Missing values:")
print(df.isnull())
print(df.isnull().sum())
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

5.Data Preprocessing

i.(Encoding categorical variables-Lebel Encoder)

```python
le = LabelEncoder()
df["Customer_Gender"] = le.fit_transform(df["Customer_Gender"])
print(df.tail())
```

ii.Feature scaling (Standard Scaler)

```python
sc = StandardScaler()
df[["Price","Discount"]] = sc.fit_transform(
    df[["Price","Discount"]])
print(df.head())
```

iii.Feature scaling (Normalization/MinMax Scaler)

```python
sc = MinMaxScaler()
df[["Price","Discount"]] = sc.fit_transform(
    df[["Price","Discount"]])
print(df.head())
```

7.Feature Engineering

```python
df["Price_category"]=df["Price"].apply(lambda x:"low price"if x<1000 else"medium price"if x<2000 else"high price")
print(df[["Price","Price_category"]].head())

df["Profit_category"]=df["Profit"].apply(lambda x:"poor"if x<500 else"average"if x<1500 else"excellent")
print(df[["Profit","Profit_category"]].tail())
```

8.Machine Learning Implementation

Import the Algorithm

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
f1_score, classification_report, confusion_matrix)

```

Train-Test Splitting




```python
# Features and Target
X = df.drop("Total_Sales", axis=1)
Y = df["Total_Sales"]

# Train Test Split
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("Y_train:", Y_train.shape)
print("Y_test :", Y_test.shape)

print(df.keys())
```

i.Linear Regression

```python
numeric_cols = X_train.select_dtypes(include=[np.number]).columns
X_train_numeric = X_train[numeric_cols]
X_test_numeric = X_test[numeric_cols]

model = LinearRegression()
model.fit(X_train_numeric, Y_train)
Y_pred = model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred))
print("MSE:", mean_squared_error(Y_test, Y_pred))
print("R2 Score:", r2_score(Y_test, Y_pred))

prediction = model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total Sales:", prediction[0])

```

ii.Ridge Regression



```python

ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train_numeric, Y_train)

Y_pred_ridge = ridge_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_ridge))
print("MSE:", mean_squared_error(Y_test, Y_pred_ridge))
print("R2 Score:", r2_score(Y_test, Y_pred_ridge))

prediction = ridge_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total Sales:", prediction[0])
```

iii.Lasso Regression

```python
lasso_model = Lasso(alpha=1.0, max_iter=10000)

lasso_model.fit(X_train_numeric, Y_train)

Y_pred_lasso = lasso_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_lasso))
print("MSE:", mean_squared_error(Y_test, Y_pred_lasso))
print("R2 Score:", r2_score(Y_test, Y_pred_lasso))

prediction = lasso_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total Sales:", prediction[0])
```

iv.Decision Tree Regressor

```python
dt_model = DecisionTreeRegressor(random_state=42)     #patterns learn

dt_model.fit(X_train_numeric, Y_train)

Y_pred_dt = dt_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_dt))
print("MSE:", mean_squared_error(Y_test, Y_pred_dt))
print("R2 Score:", r2_score(Y_test, Y_pred_dt))

prediction = dt_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total Sales:", prediction[0])
```

V.Support Vector Machine(svm)

```python
svm_model = SVR(kernel="rbf", C=1)          #curved/non linear relationship learn,model errors control

svm_model.fit(X_train_numeric, Y_train)

Y_pred_svm = svm_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_svm))
print("MSE:", mean_squared_error(Y_test, Y_pred_svm))
print("R2 Score:", r2_score(Y_test, Y_pred_svm))

prediction = svm_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total_Sales:", prediction[0])
```

Vi.Polynomial Regression

```python
X = df[["Quantity"]]
y = df["Total_Sales"]

poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

X_new = np.array([[5]])
X_new_poly = poly.transform(X_new)

y_pred = model.predict(X_new_poly)
print("Predicted Total_Sales:", y_pred[0])

X_curve = np.linspace(1, 9, 100).reshape(-1, 1)

X_curve_poly = poly.transform(X_curve)

y_curve = model.predict(X_curve_poly)

plt.scatter(X, y)
plt.plot(X_curve,y_curve,color="blue")
plt.xlabel("Quantity",color="red")
plt.ylabel("Total_Sales",color="red")
plt.title("Polynomial Regression",color="green")
plt.show()
```

Vii.Elastic Net Regresssion

```python
elastic_model = ElasticNet(alpha=1.0, l1_ratio=0.5, max_iter=10000)       #l1,l2

elastic_model.fit(X_train_numeric, Y_train)

Y_pred_elastic = elastic_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_elastic))
print("MSE:", mean_squared_error(Y_test, Y_pred_elastic))
print("R2 Score:", r2_score(Y_test, Y_pred_elastic))

prediction = elastic_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total_Sales:", prediction[0])

```

Viii.K-Nearest Neighbors (KNN) Algorithm

```python
knn_model = KNeighborsRegressor(n_neighbors=5)        #target value average predict

knn_model.fit(X_train_numeric, Y_train)

Y_pred_knn = knn_model.predict(X_test_numeric)

print("MAE:", mean_absolute_error(Y_test, Y_pred_knn))
print("MSE:", mean_squared_error(Y_test, Y_pred_knn))
print("R2 Score:", r2_score(Y_test, Y_pred_knn))

prediction = knn_model.predict(X_test_numeric.iloc[[0]])
print("Predicted Total_Sales:", prediction[0])
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X = df["Category"]
Y = df["Total_Sales"]

# Train Test Split
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

# Perform one-hot encoding
X_train = pd.get_dummies(X_train, dtype=int)
X_test = pd.get_dummies(X_test, dtype=int)

X_test = X_test.reindex(
    columns=X_train.columns, fill_value=0
)
print(" One Hot Encoding:")
print(X_train.head())

#  Linear Regression Model
model = LinearRegression()

# Train Model
model.fit(X_train, Y_train)

#  Prediction
y_pred = model.predict(X_test)

# Model Evaluation
print("\nMean Absolute Error:",
      mean_absolute_error(Y_test, y_pred))

print("Mean Squared Error:",
      mean_squared_error(Y_test, y_pred))

print("R2 Score:",
      r2_score(Y_test, y_pred))

# Predict Total Sales
prediction = model.predict(X_test)

print("Predicted Total Sales:", prediction[0])
```
